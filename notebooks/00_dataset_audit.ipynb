## Stage 0 conclusions

This is a supervised tabular regression task.

Target:
- SalePrice

Data structure:
- train.csv contains features and target.
- test.csv does not contain SalePrice and is intended for Kaggle submission only.
- sample_submission.csv contains the expected submission format.

Initial feature handling:
- Id is an identifier column and should not be used as a meaningful feature at baseline.
- Numerical features: 36
- Categorical features: 43

Missing values:
- Many missing values are present in both numerical and categorical columns.
- Some missing values likely mean absence of a property, e.g. PoolQC, Alley, Fence, FireplaceQu, Garage*, Bsmt*.
- No imputation should be fitted on the full dataset before split.
- Missing value handling must be done inside Pipeline / ColumnTransformer.

Leakage risks:
- No obvious post-sale leakage columns were found during the initial audit.
- Official Kaggle test.csv has no target and must not be used as validation.
- Local validation must be created from train.csv only.

Initial validation strategy:
- Create local train/test split from train.csv.
- Holdout test should be used only once for final local evaluation.
- During model comparison and tuning, use CV on training data only.

Main metrics:
- MAE: easy to interpret in dollars.
- RMSE: penalizes large errors more strongly.
- RMSLE: relevant because Kaggle evaluates log-scale error and expensive houses should not dominate purely by absolute price.

Possible target transformation:
- SalePrice is right-skewed.
- log1p(SalePrice) may be useful later.
- Target transformation must be done only inside the modeling workflow, e.g. TransformedTargetRegressor or CV-safe approach.
